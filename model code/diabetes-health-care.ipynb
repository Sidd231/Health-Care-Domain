```python
# remove ? at diag_1
fd = fd[fd["diag_1"] != "?"]
# remove ? at diag_2
fd = fd[fd["diag_2"] != "?"]
# remove ? at diag_2
fd = fd[fd["diag_3"] != "?"]
# remove ? at race 
fd = fd[fd["race"] != "?"]
# remove Unknown/Invalid at 
fd = fd[fd["gender"] != "Unknown/Invalid"]
# weight 
fd = fd[fd["weight"] != "?"]
# payer_code 
fd = fd[fd["payer_code"] != "?"]
```

```python
# drop max_glu_serum 
fd.drop(["A1Cresult","max_glu_serum","weight" , "medical_specialty","payer_code"], axis = 1 , inplace = True)
# metformin-pioglitazone,
fd.drop(["metformin-pioglitazone","metformin-rosiglitazone","glimepiride-pioglitazone"
        ,"glipizide-metformin","glyburide-metformin","insulin","citoglipton"
        ,"tolazamide","miglitol","troglitazone","troglitazone","acetohexamide"
         ,"examide","tolbutamide","chlorpropamide","examide","change"
        ], axis = 1 , inplace = True)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoders = {}
col = fd.columns
for c in col:
    le = LabelEncoder()
    fd[c] = le.fit_transform(fd[c])  # Male/Female → 0/1
    encoders[c] = le
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = fd.drop("readmitted" , axis = 1)
y = fd["readmitted"]
```

```python
 X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
import xgboost as xg
```

```python
model = xg.XGBClassifier(
    n_estimators = 150,
    learning_rate = 0.1,
    subsample = 0.8
)
```

```python
model.fit(X_train , y_train)
```

```python
predictions = model.predict(X_test)       # Class labels (0, 1, 2...)
probabilities = model.predict_proba(X_test) 
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Accuracy
print("Accuracy:", accuracy_score(y_test, predictions))

# Detailed report (precision, recall, f1-score)
print(classification_report(y_test, predictions))

# Confusion matrix
print(confusion_matrix(y_test, predictions))
```

```python
import matplotlib.pyplot as plt
import xgboost as xgb

# Plot feature importance
xgb.plot_importance(model, max_num_features=10)
plt.show()

# Get importance values as DataFrame
importance = pd.DataFrame({
    'feature': model.feature_names_in_,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)
print(importance.head(10))
```
